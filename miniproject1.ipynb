```python

from collections import deque

# SNAKE AND LADDER USING BFS
# DSA Mini Project

BOARD_SIZE = 100

# Snakes: starting square -> ending square
snakes = {
    99: 54,
    70: 55,
    52: 42,
    25: 2,
    95: 72,
    89: 68,
    64: 60,
    62: 19,
    49: 11,
    46: 25
}

# Ladders: starting square -> ending square
ladders = {
    6: 27,
    11: 40,
    17: 69,
    20: 38,
    28: 84,
    36: 44,
    51: 67,
    63: 81,
    71: 91,
    80: 100
}

# Combine snakes and ladders
jumps = {**snakes, **ladders}


def bfs_snake_ladder():
    # Queue stores squares to visit
    queue = deque([1])

    # Track visited squares
    visited = {1}

    # Store parent and dice information for path reconstruction
    parent = {1: None}
    move_info = {}

    while queue:
        current = queue.popleft()

        # Destination reached
        if current == BOARD_SIZE:
            break

        # Try every possible dice result
        for dice in range(1, 7):
            landing = current + dice

            # Do not allow moves beyond square 100
            if landing > BOARD_SIZE:
                continue

            # Apply snake or ladder
            destination = jumps.get(landing, landing)

            if destination not in visited:
                visited.add(destination)
                parent[destination] = current
                move_info[destination] = (dice, landing)

                queue.append(destination)

    # Check whether the destination is reachable
    if BOARD_SIZE not in parent:
        print("No solution found.")
        return

    # Reconstruct the shortest path
    path = []
    current = BOARD_SIZE

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    # Display project results
    print("=" * 55)
    print("       SNAKE AND LADDER USING BFS")
    print("=" * 55)

    print("\nStarting Square:", 1)
    print("Destination Square:", BOARD_SIZE)
    print("Minimum Dice Throws:", len(path) - 1)

    print("\nShortest Path:")
    print(" -> ".join(map(str, path)))

    print("\nDetailed Moves:")

    for i in range(1, len(path)):
        previous = path[i - 1]
        destination = path[i]
        dice, landing = move_info[destination]

        print(
            f"Square {previous} -- Dice: {dice} --> "
            f"Square {landing}",
            end=""
        )

        if landing in ladders:
            print(f" [LADDER to {destination}]")
        elif landing in snakes:
            print(f" [SNAKE to {destination}]")
        else:
            print(f" [Reached {destination}]")

    print("\nSnakes:", snakes)
    print("Ladders:", ladders)
    print("\nBFS completed successfully!")


# Run the project
bfs_snake_ladder()
```
